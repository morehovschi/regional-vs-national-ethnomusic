## Step 1: Preprocessing
#### 1.1 Borsan dataset: filter out generated MIDI files that the IDyOM implementation cannot digest.
In the process of automatically converting the Borsan dataset from MEI to MIDI, some metadata is probably
added that IDyOM does not expect, leading to an interrupting of the analysis loop.
Filter out those problematic files.

```python
target_dir = "inter-data/borsan-midi"
fnames = os.listdir( target_dir )
fnames = [ f"{target_dir}/{fname}" for fname in fnames ]
```

```python
bad_files = []

for midi_file in fnames:
    print(f"Processing {midi_file}...")
    
    # Create a string buffer to capture output
    output_buffer = io.StringIO()
    
    try:
        with redirect_stdout(output_buffer):
            my_experiment = IDyOMExperiment(test_dataset_path=midi_file,
                                          pretrain_dataset_path='data/dissimilarities-22/MIDI_files/FS_MIDI/FS_MIDI_SLOVENIA/')
            
            my_experiment.set_parameters(target_viewpoints=['cpitch', 'onset'],
                                       source_viewpoints=['cpitch', 'onset'],
                                       models=':both',
                                       k=1,
                                       detail=3)
            my_experiment.run()
            
        # Get the captured output
        output = output_buffer.getvalue()
        
        # Check for error indicators in the output
        if "Type HELP for debugger help" in output or "NO-APPLICABLE-METHOD" in output:
            bad_files.append(midi_file)
            print(f"  Error detected in {midi_file}")
            
    except Exception as e:
        bad_files.append(midi_file)
        print(f"  Exception occurred with {midi_file}: {str(e)}")
    
    output_buffer.close()

print("\nFiles with errors:")
for bad_file in bad_files:
    print(bad_file)
```

```python
def parse_experiment_output(output_text):
    lines = output_text.split('\n')
    
    good_files = []
    bad_files = []
    current_file = None
    
    # Patterns that indicate a bad run
    error_patterns = [
        "Type HELP for debugger help",
        "debugger invoked on",
        "invalid number of arguments",
        "SB-INT:SIMPLE-PROGRAM-ERROR",
        "restarts (invokable by number",
        "NO-APPLICABLE-METHOD"
    ]
    
    for line in lines:
        # Detect when a new file is being processed
        if line.startswith('Processing '):
            current_file = line.split('Processing ')[1].strip()
            is_bad = False
        
        # Check for error patterns
        if current_file and any(error in line for error in error_patterns):
            is_bad = True
        
        # Detect the end of processing (either good or bad)
        if current_file and ("Written PPM* model to" in line or "*  " in line):
            if is_bad:
                bad_files.append(current_file)
            else:
                good_files.append(current_file)
            current_file = None
    
    # Handle case where processing ended without the usual markers
    if current_file:
        bad_files.append(current_file)
    
    return good_files, bad_files
```

## 2. Extract features from the two datasets
#### 2.1 Extract from the Mihelač dataset, by country
From each one of the 22 countries in the Mihelač dataset, extract a vector of 16 features. These
represent the surprise values for every note averaged over time for each one of 16 features that
the model considers.

```python
# define the viewpoints (as source-target viewpoint pairs)

source_target_pairs = [
  # ( source, target )
    ( [ "cpitch" ], [ "cpitch" ] ),
    ( [ "cpitch", "dur" ], [ "cpitch", "dur" ] ),
    # ( [ "dur", "dur-ratio" ], [ "dur" ] ), BROKEN
    ( [ "dur" ], [ "dur" ] ),  # using this instead of dur-ratio
    ( [ "cpitch", "inscale" ], [ "cpitch" ] ),
    ( [ "cpitch", "cpint" ], [ "cpitch" ] ),
    ( [ "cpitch", "cpint-size" ], [ "cpitch" ] ),
    ( [ "cpitch", "cpitch-class" ], [ "cpitch" ] ),
    ( [ "cpitch", "cpcint" ], [ "cpitch" ] ),
    ( [ "cpitch", "cpcint-size" ], [ "cpitch" ] ),
    ( [ "cpitch", "contour" ], [ "cpitch" ] ),
    ( [ "cpitch", "newcontour" ], [ "cpitch" ] ),
    ( [ "cpitch", "cpintfref" ], [ "cpitch" ] ),
    ( [ "cpitch", "cpintfip" ], [ "cpitch" ] ),
    ( [ "cpitch", "tessitura" ], [ "cpitch" ] ),
    ( [ "onset", "ioi-ratio" ], [ "onset" ] ),
    ( [ "cpitch", "onset", "ioi-ratio" ], [ "cpitch", "onset" ] )
]
```

```python
!rm -r experiment_history
!mkdir experiment_history
```

```python
for i, country_name in enumerate( country_names ):
    print( f"Calculating features for country {i+1}/{len(country_names)}: {country_name}." )
    for i, ( source_viewpoints, target_viewpoints ) in enumerate( source_target_pairs ):
        print( f"Calculating feature {i+1}/{len(source_target_pairs)}: {source_viewpoints}")
        
        prev_experiment_experiments = set( os.listdir( "experiment_history" ) )
        
        my_experiment = IDyOMExperiment(test_dataset_path=f'inter-data/{country_name.lower()}-fs-cfs/',
                                        pretrain_dataset_path='inter-data/mihelac-borsan-fs-all/')

        print( source_viewpoints, target_viewpoints )
        my_experiment.set_parameters(
            target_viewpoints=target_viewpoints,
            source_viewpoints=source_viewpoints,
            models=":both",
            k=1,
            detail=3,
        )

        my_experiment.run()
        
        new_experiment_name =\
            list( ( set( os.listdir( "experiment_history" ) ) - prev_experiment_experiments ) )[ 0 ]
        
        !mv experiment_history/{new_experiment_name} experiment_history/{country_name.lower()}_viewpoint{i+1}
```
